```python
from pydantic import BaseModel
from langchain_core.prompts import ChatPromptTemplate

from langchain_ollama import ChatOllama

class Answer(BaseModel):
    message: str
    mood: str
    confidence: float

llm = ChatOllama(model="gemma:2b", temperature=0)

structured_llm = llm.with_structured_output(Answer)

result = structured_llm.invoke(
    "I'm so happy that I got my first job!"
)
result
```

```python

from langchain_ollama import ChatOllama

from langchain_core.prompts import ChatPromptTemplate
from pydantic import BaseModel, Field

class MedicineInfo(BaseModel):
  medicine_name : str = Field(description="medicine name")
  strength : str  = Field(description="strength of medicine")

llm =ChatOllama(model="gemma:2b")

prompt = ChatPromptTemplate.from_messages([
  ("system", "extract medical information from the users text"),
  ("human" , "{text}")
])



structured_ouput = llm.with_structured_ouput(MedicineInfo)



chain = prompt | structured_ouput


chain.invoke({
  "text": "paracetamol"
})

```
